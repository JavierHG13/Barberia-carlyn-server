# Clasificacion: prediccion de inasistencias

Esta libreta compara varios modelos de clasificacion para predecir si una cita puede terminar como **no asistio** o **cancelada**.

El objetivo no es reemplazar la operacion del negocio, sino apoyar al administrador para detectar citas con mayor riesgo y tomar acciones antes de perder el horario.

## 1. Importacion de librerias

Se importan las librerias para:

- cargar datos desde PostgreSQL
- preparar variables categoricas y numericas
- entrenar diferentes modelos
- aplicar validacion cruzada
- obtener metricas como accuracy, precision, recall y F1

```python
from pathlib import Path
import os
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import psycopg2

from sklearn.compose import ColumnTransformer
from sklearn.ensemble import GradientBoostingClassifier, RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report
from sklearn.model_selection import StratifiedKFold, cross_val_score, cross_validate, train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.tree import DecisionTreeClassifier

sns.set_theme(style="whitegrid")
```

## 2. Conexion con la base de datos

La libreta usa las credenciales del archivo `.env` del backend. El dataset se toma desde `analitica.ml_citas_dataset`, que es la tabla preparada para evidencia de mineria de datos.

```python
SERVER_ROOT = Path.cwd()
if SERVER_ROOT.name == "notebooks":
    SERVER_ROOT = SERVER_ROOT.parents[1]
elif SERVER_ROOT.name == "ml":
    SERVER_ROOT = SERVER_ROOT.parent

ARTIFACT_DIR = SERVER_ROOT / "ml" / "artifacts"
ARTIFACT_DIR.mkdir(parents=True, exist_ok=True)


def load_env(path):
    env = {}
    if not Path(path).exists():
        return env
    for raw in Path(path).read_text(encoding="utf-8").splitlines():
        line = raw.strip()
        if not line or line.startswith("#") or "=" not in line:
            continue
        key, value = line.split("=", 1)
        env[key.strip()] = value.strip().strip('"').strip("'")
    return env


env = {**load_env(SERVER_ROOT / ".env"), **os.environ}

conn = psycopg2.connect(
    host=env.get("DB_HOST", "localhost"),
    port=int(env.get("DB_PORT", 5432)),
    user=env.get("DB_USER", "postgres"),
    password=env.get("DB_PASSWORD", ""),
    dbname=env.get("DB_NAME", "db_barberia"),
)

query = "SELECT * FROM analitica.ml_citas_dataset ORDER BY id"
df = pd.read_sql_query(query, conn)
conn.close()

df["fecha"] = pd.to_datetime(df["fecha"])
df["hora"] = pd.to_datetime(df["hora"].astype(str), format="%H:%M:%S", errors="coerce").dt.hour

df.head()
```

## 3. Revision inicial del dataset

Se revisa el tamano del dataset y la distribucion de estados. Las citas pendientes se excluyen del entrenamiento porque todavia no tienen resultado real.

```python
print("Filas y columnas:", df.shape)

df[[
    "id", "cliente_nombre", "local_nombre", "servicio_nombre", "barbero_nombre",
    "fecha", "hora", "precio", "duracion", "estado_cita"
]].head(10)
```

```python
df["estado_cita"].value_counts().rename_axis("estado_cita").reset_index(name="total")
```

## 4. Construccion de la variable objetivo

La variable dependiente `y` se llama `target_no_show`.

- `0`: la cita fue asistida.
- `1`: la cita fue cancelada o termino como no-show.

Esta conversion permite convertir el problema del negocio en un problema de clasificacion binaria.

```python
data = df[df["estado_cita"] != "pendiente"].copy()
data["target_no_show"] = data["estado_cita"].isin(["no_show", "cancelada"]).astype(int)

data[["estado_cita", "target_no_show"]].value_counts().reset_index(name="total")
```

## 5. Variables independientes X y variable objetivo y

Variables `X` usadas para explicar el riesgo:

| Tipo | Variables |
|---|---|
| Sucursal y agenda | local, dia, semana, hora |
| Servicio | nombre del servicio, precio, duracion |
| Barbero | barbero asignado |
| Cliente | frecuencia, recencia, gasto total, tasa previa de no-show |
| Operacion | recordatorio enviado, canal de reserva |

Variable `y`:

| Variable | Significado |
|---|---|
| target_no_show | 1 si hubo cancelacion/no-show, 0 si asistio |

```python
features = [
    "local_nombre", "servicio_nombre", "barbero_nombre",
    "dia_semana", "semana", "hora", "precio", "duracion",
    "recordatorio_enviado", "frecuencia_cliente", "recencia_dias",
    "gasto_total_cliente", "no_show_rate_cliente", "canal",
]

categorical_features = ["local_nombre", "servicio_nombre", "barbero_nombre", "canal"]
numeric_features = [column for column in features if column not in categorical_features]

X = data[features]
y = data["target_no_show"]

print("X:", X.shape)
print("y:", y.shape)
X.head()
```

## 6. Division de datos

Se separa el dataset en entrenamiento y prueba.

- 75% para entrenamiento
- 25% para prueba

Se usa `stratify=y` para conservar la proporcion de citas con riesgo y sin riesgo en ambos conjuntos.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y,
)

print("Entrenamiento:", X_train.shape)
print("Prueba:", X_test.shape)
```

## 7. Preprocesamiento

Los modelos necesitan datos numericos. Por eso se aplica:

- `OneHotEncoder` para variables categoricas como sucursal, servicio, barbero y canal.
- `StandardScaler` para variables numericas como precio, duracion, frecuencia y recencia.

Este preprocesamiento se integra en un `Pipeline` para que cada modelo reciba los datos de la misma forma.

```python
preprocess = ColumnTransformer([
    ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features),
    ("num", StandardScaler(), numeric_features),
])


def build_model(estimator):
    return Pipeline([
        ("prep", preprocess),
        ("model", estimator),
    ])
```

## 8. Modelos a comparar

Se comparan cinco modelos:

| Modelo | Por que se usa |
|---|---|
| LogisticRegression | Base estadistica simple e interpretable |
| KNeighborsClassifier | Clasifica por similitud con clientes/citas cercanas |
| DecisionTreeClassifier | Facil de explicar con reglas |
| RandomForestClassifier | Combina muchos arboles y suele ser estable |
| GradientBoostingClassifier | Ensamble secuencial que corrige errores previos |

```python
modelos = {
    "LogisticRegression": build_model(LogisticRegression(max_iter=1200, class_weight="balanced")),
    "KNeighborsClassifier": build_model(KNeighborsClassifier(n_neighbors=9)),
    "DecisionTreeClassifier": build_model(DecisionTreeClassifier(max_depth=6, min_samples_leaf=10, class_weight="balanced", random_state=42)),
    "RandomForestClassifier": build_model(RandomForestClassifier(n_estimators=160, max_depth=9, min_samples_leaf=6, class_weight="balanced", random_state=42)),
    "GradientBoostingClassifier": build_model(GradientBoostingClassifier(n_estimators=120, learning_rate=0.06, max_depth=3, random_state=42)),
}

modelos
```

## 9. Validacion cruzada

Se utiliza validacion cruzada estratificada con 6 folds.

Esto significa que el dataset se divide en 6 partes. El modelo entrena varias veces usando diferentes particiones y despues se promedian los resultados.

La metrica principal que se muestra es `accuracy`, pero tambien se calculan `precision`, `recall` y `f1` para tener una comparacion mas completa.

```python
cv = StratifiedKFold(n_splits=6, shuffle=True, random_state=42)

resultados = []
accuracy_por_fold = {}

for nombre, modelo in modelos.items():
    scores_accuracy = cross_val_score(modelo, X, y, cv=cv, scoring="accuracy")
    scores = cross_validate(
        modelo,
        X,
        y,
        cv=cv,
        scoring={
            "accuracy": "accuracy",
            "precision": "precision",
            "recall": "recall",
            "f1": "f1",
        },
    )

    accuracy_por_fold[nombre] = scores_accuracy
    resultados.append({
        "Modelo": nombre,
        "Accuracy Promedio": scores["test_accuracy"].mean(),
        "Desviacion Accuracy": scores["test_accuracy"].std(),
        "Precision Promedio": scores["test_precision"].mean(),
        "Recall Promedio": scores["test_recall"].mean(),
        "F1 Promedio": scores["test_f1"].mean(),
    })

tabla_metricas = pd.DataFrame(resultados).sort_values(
    by=["Accuracy Promedio", "F1 Promedio"],
    ascending=False,
).reset_index(drop=True)

tabla_metricas
```

## 10. Accuracy por fold

Esta tabla muestra el accuracy obtenido por cada modelo en cada fold de la validacion cruzada.

```python
tabla_folds = pd.DataFrame(accuracy_por_fold).T
tabla_folds.columns = [f"Fold {i}" for i in range(1, tabla_folds.shape[1] + 1)]
tabla_folds["Accuracy Promedio"] = tabla_folds.mean(axis=1)
tabla_folds
```

## 11. Seleccion del mejor modelo

Para elegir el mejor modelo se ordena por `Accuracy Promedio`. En caso de empate, se usa `F1 Promedio` como segundo criterio.

En un caso real de no-show tambien es importante revisar `Recall`, porque perder una cita riesgosa puede afectar directamente la agenda.

```python
mejor_modelo_nombre = tabla_metricas.iloc[0]["Modelo"]
mejor_modelo = modelos[mejor_modelo_nombre]

print("Mejor modelo seleccionado:", mejor_modelo_nombre)
tabla_metricas.head(1)
```

## 12. Entrenamiento final y accuracy en prueba

Despues de elegir el mejor modelo con validacion cruzada, se entrena con el conjunto de entrenamiento y se evalua contra el conjunto de prueba.

```python
mejor_modelo.fit(X_train, y_train)
y_pred = mejor_modelo.predict(X_test)

metricas_prueba = pd.DataFrame([{
    "Modelo": mejor_modelo_nombre,
    "Accuracy": accuracy_score(y_test, y_pred),
    "Precision": precision_score(y_test, y_pred, zero_division=0),
    "Recall": recall_score(y_test, y_pred, zero_division=0),
    "F1": f1_score(y_test, y_pred, zero_division=0),
}])

metricas_prueba
```

## 13. Matriz de confusion

La matriz de confusion permite revisar aciertos y errores:

- Citas sin riesgo clasificadas correctamente.
- Citas con riesgo detectadas correctamente.
- Falsas alarmas.
- Citas riesgosas no detectadas.

```python
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(5, 4))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Sin riesgo", "Riesgo"],
    yticklabels=["Sin riesgo", "Riesgo"],
)
plt.xlabel("Prediccion")
plt.ylabel("Valor real")
plt.title(f"Matriz de confusion - {mejor_modelo_nombre}")
plt.show()

print(classification_report(y_test, y_pred, target_names=["Sin riesgo", "Riesgo"], zero_division=0))
```

## 14. Aplicacion del modelo a citas pendientes

Con el modelo seleccionado se calcula la probabilidad de riesgo para citas pendientes. Esto genera una lista priorizada para el administrador.

```python
citas_pendientes = df[df["estado_cita"] == "pendiente"].copy()

if citas_pendientes.empty:
    citas_pendientes = df.sort_values("fecha", ascending=False).head(80).copy()

citas_pendientes["riesgo_no_show"] = mejor_modelo.predict_proba(citas_pendientes[features])[:, 1]

citas_riesgo = citas_pendientes.sort_values("riesgo_no_show", ascending=False).head(15)

citas_riesgo[[
    "fecha", "hora", "cliente_nombre", "servicio_nombre", "barbero_nombre",
    "local_nombre", "riesgo_no_show"
]]
```

## 15. Guardado del modelo

Se guarda el mejor modelo entrenado en `.joblib`. Este archivo puede cargarse despues para generar predicciones sin volver a entrenar.

```python
artifact_path = ARTIFACT_DIR / "notebook_no_show_best_classifier.joblib"
joblib.dump({
    "modelo": mejor_modelo,
    "nombre_modelo": mejor_modelo_nombre,
    "tabla_metricas": tabla_metricas,
    "features": features,
    "categorical_features": categorical_features,
    "numeric_features": numeric_features,
}, artifact_path)

artifact_path
```
